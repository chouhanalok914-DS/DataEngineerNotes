```python
# 7. Find all vehicles with top_speed_kmh greater than 180
```

```python
(df.createOrReplaceTempView("df_view"))
```

```python
query = """
          select brand , model, top_speed_kmh
          from df_view
          where top_speed_kmh > 180
        """
```

```python
spark.sql(query).show(5)
```

```python
# 7. Filter rows where torque_nm is null.
```

```python
(
    df
    .select("brand","model","torque_nm")
    .filter(psf.expr("torque_nm is null"))
    .show(5)
)
```

```python
# 9 retrieve vehicles where range_km is between 250 and 300.
```

```python
query = """
        select brand , model, range_km
        from df_view
        where range_km between 250 and 300
        order by range_km
        """
```

```python
spark.sql(query).show(5)
```
